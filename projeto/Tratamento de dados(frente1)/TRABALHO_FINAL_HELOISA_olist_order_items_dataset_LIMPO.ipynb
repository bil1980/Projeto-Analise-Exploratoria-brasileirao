```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
```

```python


arquivos = files.upload()
df = pd.read_csv("olist_order_items_dataset.csv")
df_original = df.copy(0)
```

```python
df.head()
df.info()
df.describe(include="all")
```

```python
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df.duplicated().sum()

df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df["shipping_limit_date"] = pd.to_datetime(
    df["shipping_limit_date"], errors="coerce"
)
```

```python
#Checagem de numeros negativos
colunas_numericas = df.select_dtypes(include="number").columns


for coluna in colunas_numericas:
    negativos = df[df[coluna] < 0]

    if not negativos.empty:
        print(f"\nColuna: {coluna}")
        print(f"Quantidade de negativos: {len(negativos)}")
        display(negativos)
```

```python
#checagem de outliers
for coluna in colunas_numericas:

    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = df[
        (df[coluna] < limite_inferior) |
        (df[coluna] > limite_superior)
    ]

    print(f"\nColuna: {coluna}")
    print(f"Possíveis outliers: {len(outliers)}")
```

```python
df.to_csv("olist_order_items_dataset_limpo.csv", index=False)

df.to_csv(
    "olist_order_items_dataset_limpo.csv",
    index=False,
    sep=";",
    encoding="utf-8-sig"
)
```
